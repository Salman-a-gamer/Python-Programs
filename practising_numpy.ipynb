```python
# import statistics
# import random
import numpy as np
```

```python
# create an array
array1d = np.array([1,2,3,4,5])
array2d = np.array([[1,2,3], [4,5,6], [7,8,':']])
print(array2d)
```

```python
# get the dimension of the array.
print(array2d.ndim)
array1d
```

```python
# get the shape of the array.
print(array1d.shape)
array2d.shape
```

```python
# data type of an element
print(array1d.dtype)
print(array2d.dtype)
```

```python
# get the size in bytes of an element
print(array1d.itemsize)
print(array2d.itemsize)
```

```python
# get the no. of elements in each array
print(array1d.size)
print(array2d.size)
```

```python
#get total size of array
print(array1d.nbytes)
array2d.nbytes
```

```python
# accessing elements...
# arrayname[row  ,  col ]
print(array1d[:]) # get entire elements
print(array2d[:, 2]) # get all rows, with column 1 -> getting a specific column
```
